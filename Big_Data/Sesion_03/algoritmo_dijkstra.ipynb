# Algoritmo Dikstra

## 1. ¿Qué es el algoritmo de Dijkstra?

El **algoritmo de Dijkstra** sirve para encontrar el **camino más corto desde un punto de partida hasta los demás puntos de un grafo**.

Por ejemplo, imagina estas ciudades:

```text
        4
   A -------- B
   |          |
  2|          |1
   |          |
   C -------- D
        3
```

Los números representan el **coste o distancia** entre los puntos.

Si queremos saber cuál es el camino más corto desde `A` hasta `D`, tenemos varias posibilidades:

* `A → B → D` = 4 + 1 = **5**
* `A → C → D` = 2 + 3 = **5**

Por tanto, la distancia mínima desde `A` hasta `D` es **5**.

---

## 2. La idea principal de Dijkstra

Dijkstra va tomando decisiones de forma progresiva.

Podemos imaginar que tenemos una lista de distancias conocidas:

| Nodo | Distancia desde A |
| ---- | ----------------: |
| A    |                 0 |
| B    |                 ∞ |
| C    |                 ∞ |
| D    |                 ∞ |

Al principio:

* Estamos en `A`, por eso su distancia es `0`.
* Todavía no sabemos cuánto cuesta llegar a los demás, así que ponemos `∞`.

Después visitamos `A` y comprobamos sus vecinos:

```text
A → B = 4
A → C = 2
```

Actualizamos:

| Nodo | Distancia |
| ---- | --------: |
| A    |         0 |
| B    |         4 |
| C    |         2 |
| D    |         ∞ |

Ahora Dijkstra escoge el nodo **no visitado con menor distancia**.

Es `C`, porque cuesta `2`.

Desde `C` podemos llegar a `D`:

```text
A → C → D
0 + 2 + 3 = 5
```

Entonces:

| Nodo | Distancia |
| ---- | --------: |
| A    |         0 |
| B    |         4 |
| C    |         2 |
| D    |         5 |

Después continúa hasta que ha encontrado las distancias mínimas.

### La idea

Dijkstra repite básicamente estos pasos:

1. Elegir el nodo más cercano que todavía no hemos procesado.
2. Mirar sus vecinos.
3. Comprobar si podemos llegar a ellos mediante un camino más corto.
4. Actualizar las distancias si encontramos un camino mejor.
5. Repetir.

Esto último se suele llamar **relajación de las aristas**.

---

# 3. Una implementación sencilla en Python

Para aprender, podemos empezar con una versión sencilla utilizando un diccionario.

```python
def dijkstra(grafo, inicio):

    distancias = {}
    visitados = set()

    # Al principio todas las distancias son infinitas
    for nodo in grafo:
        distancias[nodo] = float("inf")

    # La distancia del inicio hasta sí mismo es 0
    distancias[inicio] = 0

    while len(visitados) < len(grafo):

        # Buscar el nodo no visitado con menor distancia
        nodo_actual = None

        for nodo in grafo:
            if nodo not in visitados:
                if nodo_actual is None or distancias[nodo] < distancias[nodo_actual]:
                    nodo_actual = nodo

        # Marcarlo como visitado
        visitados.add(nodo_actual)

        # Comprobar sus vecinos
        for vecino, distancia in grafo[nodo_actual].items():

            nueva_distancia = distancias[nodo_actual] + distancia

            if nueva_distancia < distancias[vecino]:
                distancias[vecino] = nueva_distancia

    return distancias
```
Vamos a desmontarlo.

# 4. Representamos el grafo

Podemos representar nuestro ejemplo mediante un diccionario:

```python
grafo = {
    "A": {"B": 4, "C": 2},
    "B": {"A": 4, "D": 1},
    "C": {"A": 2, "D": 3},
    "D": {"B": 1, "C": 3}
}
```

Por ejemplo:

```python
"A": {"B": 4, "C": 2}
```

significa:

> Desde A puedo ir a B con un coste de 4 y a C con un coste de 2.

Visualmente:

```text
        4
   A -------- B
   |          |
  2|          |1
   |          |
   C -------- D
        3
```

# 5. Ejecutamos Dijkstra

Ahora simplemente:

```python
resultado = dijkstra(grafo, "A")

print(resultado)
```

Obtendremos:

```text
{'A': 0, 'B': 4, 'C': 2, 'D': 5}
```

Esto significa:

```text
Desde A:

A → A = 0
A → B = 4
A → C = 2
A → D = 5
```

Y aquí hay algo interesante.

Aunque directamente tenemos:

```text
A → B = 4
```

para llegar a `D` podemos hacer:

```text
A → C → D
```

que cuesta:

```text
2 + 3 = 5
```

Por eso:

```text
distancia A → D = 5
```

# 6. Una versión todavía más fácil de entender

### Celda 1 — Definir el grafo

```python
grafo = {
    "A": {"B": 4, "C": 2},
    "B": {"A": 4, "D": 1},
    "C": {"A": 2, "D": 3},
    "D": {"B": 1, "C": 3}
}
```

### Celda 2 — Algoritmo

```python
def dijkstra(grafo, inicio):

    distancias = {}
    visitados = set()

    for nodo in grafo:
        distancias[nodo] = float("inf")

    distancias[inicio] = 0

    while len(visitados) < len(grafo):

        nodo_actual = None

        for nodo in grafo:
            if nodo not in visitados:
                if nodo_actual is None or distancias[nodo] < distancias[nodo_actual]:
                    nodo_actual = nodo

        visitados.add(nodo_actual)

        for vecino, distancia in grafo[nodo_actual].items():

            nueva_distancia = distancias[nodo_actual] + distancia

            if nueva_distancia < distancias[vecino]:
                distancias[vecino] = nueva_distancia

    return distancias
```

### Celda 3 — Ejecutarlo

```python
resultado = dijkstra(grafo, "A")

print(resultado)
```

Resultado:

```text
{'A': 0, 'B': 4, 'C': 2, 'D': 5}
```

## 7. ¿Qué significa `float("inf")`?

Esta parte es importante:

```python
float("inf")
```

significa **infinito**.

Lo utilizamos porque al principio no conocemos las distancias:

```python
A = 0
B = ∞
C = ∞
D = ∞
```

Luego Dijkstra va sustituyendo esos infinitos por distancias reales cuando encuentra caminos.

---

## 8. ¿Y qué hace esta parte?

Esta es probablemente la parte más importante del algoritmo:

```python
nueva_distancia = distancias[nodo_actual] + distancia

if nueva_distancia < distancias[vecino]:
    distancias[vecino] = nueva_distancia
```

Por ejemplo, estamos en `C`.

Tenemos:

```text
A → C = 2
C → D = 3
```

Entonces:

```python
nueva_distancia = 2 + 3
```

Resultado:

```text
5
```

Si anteriormente teníamos:

```text
D = ∞
```

hacemos:

```text
5 < ∞
```

Por tanto:

```text
D = 5
```

Y si posteriormente encontrásemos otro camino que costase `4`, haríamos:

```text
4 < 5
```

y actualizaríamos:

```text
D = 4
```

Eso es la **relajación**.

## 10. Una cosa importante para tu curso

Dijkstra funciona correctamente cuando los pesos de las conexiones son **positivos o cero**.

Por ejemplo:

```text
A → B = 5
B → C = 2
C → D = 8
```

✅ Correcto.

Pero si tenemos:

```text
A → B = -5
```

❌ Dijkstra no es el algoritmo adecuado.

Para grafos que pueden tener pesos negativos existe, por ejemplo, **Bellman-Ford**.
