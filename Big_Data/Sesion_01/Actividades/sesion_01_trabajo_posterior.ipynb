## Trabajo posterior
Análisis de diferentes operaciones sobre el dataset Sample Sales Data
utilizando pandas.

```python
import pandas as pd
import time
```

- ### Leer el CSV y medir el tiempo

```python
inicio = time.time()
datos = pd.read_csv("sales_data_sample.csv", encoding="latin1")
fin = time.time()

tiempo_lectura = fin - inicio

print(f"Tiempo de lectura: {tiempo_lectura:.6f} segundos")
```

```python
datos.head()  #obtiene las 5 primera líneas ordenadas
```

- ### Medir tiempos de ordenación

Medir tiempos de ordenación de una columna, en este caso de la columna `SALES`

```python
inicio = time.time()
datos_ordenados = datos.sort_values("SALES")
fin = time.time()

tiempo_ordenacion = fin - inicio

print(f"Tiempo de ordenación: {tiempo_ordenacion:.6f} segundos")
```

```python
datos_ordenados[["SALES"]].head() # sort los ordena siempre de menor a mayor por defecto
```

```python
datos_ordenados[["SALES"]].tail() # así se obitienen los valores más grandes ordenados
```

- ### Filtrado por una condición

Selecciono las ventas superiores a 5000

```python
datos_filtrados = datos[datos["SALES"] > 5000] #  solamente con las filas donde SALES sea mayor que 5000

datos_filtrados.head()
```

Ahora medir el tiempo de filtrado

```python
inicio = time.time()
datos_filtrados = datos[datos["SALES"] > 5000]
fin = time.time()

tiempo_filtrado = fin - inicio

print(f"Tiempo de filtrado: {tiempo_filtrado:.6f} segundos. Para SALES mayor a 5000.")
```

- ### Medir el tiempo de agrupar y calcular totales

Agrupo por el total de ventas de cada país:

```python
totales_pais = datos.groupby("COUNTRY")["SALES"].sum()  # Agrupar con groupby por País seleccionando la columnna SALES de cada uno y sumando el total de SALES en cada país

totales_pais
```

Medir tiempos

```python
inicio = time.time()
totales_pais = datos.groupby("COUNTRY")["SALES"].sum()
fin = time.time()

tiempo_agrupacion = fin - inicio

print(f"Tiempo de agrupación y cálculo de totales: {tiempo_agrupacion:.6f} segundos.")
```

## Resultados

| Operación | Condición / criterio | Tiempo (segundos) |
|---|---|---:|
| Lectura del archivo | `sales_data_sample.csv` | 0,027978 |
| Ordenación | Columna `SALES` | 0,009114 |
| Filtrado | `SALES > 5000` | 0,002331 |
| Agrupación y totales | Agrupado por `COUNTRY`, suma de `SALES` | 0,004610 |

## Complejidad de las operaciones

### 1. Lectura del archivo

La lectura del archivo se puede considerar una operación de complejidad
O(n), ya que es necesario recorrer los datos del archivo para cargarlos
en el DataFrame.

El dataset contiene 2.823 registros y la lectura tardó
0,027978 segundos.

### 2. Ordenación

La ordenación por la columna `SALES` tiene una complejidad O(n log n).
Esta complejidad es superior a la lineal porque, al aumentar el número de
registros, el trabajo necesario para ordenar los datos también aumenta.

La ordenación de los 2.823 registros tardó 0,009114
segundos.

### 3. Filtrado

El filtrado por una condición se puede considerar una operación de
complejidad O(n), ya que es necesario comprobar los registros para
determinar cuáles cumplen la condición indicada.

Utilizo la condición `SALES > 5000` sobre los 2.823
registros del dataset. El filtrado tardó 0,002331 segundos.

### 4. Agrupación y cálculo de totales

La agrupación por la columna `COUNTRY` y el cálculo de la suma de `SALES`
se pueden considerar una operación de complejidad O(n), ya que es necesario
procesar los registros del dataset para realizar la agrupación y calcular
los totales.

Se procesaron 2.823 registros, agrupándolos por país y
sumando sus ventas. La operación tardó 0,004610 segundos.

## Conclusión

Los resultados muestran que el tiempo de ejecución depende de la operación
realizada y de la cantidad de datos procesados.

La lectura, el filtrado y la agrupación se pueden considerar operaciones
de complejidad O(n), mientras que la ordenación presenta una complejidad
O(n log n).

En este dataset, formado por 2.823 registros, los tiempos obtenidos son
pequeños. Sin embargo, al aumentar considerablemente el número de datos,
las diferencias entre las complejidades pueden ser más importantes.

Por esto creo que conocer la complejidad algorítmica permite anticipar cómo
puede comportarse una operación cuando se trabaja con conjuntos de datos
mucho mayores.
