```python
import time
import random

# Generar lista de números aleatorios
def generar_lista(n):
    return [random.randint(1, 10000) for _ in range(n)]

# Medir tiempo de ordenación
def medir_tiempo(lista):
    inicio = time.time()
    sorted(lista)  # Ordenación nativa de Python (Timsort)
    fin = time.time()
    return fin - inicio

# Probar con diferentes tamaños
tamanos = [1000, 10000, 100000, 1000000]

print("Tamaño\t\tTiempo (segundos)")
print("-" * 35)

for n in tamanos:
    lista = generar_lista(n)
    tiempo = medir_tiempo(lista)
    print(f"{n:,}\t\t{tiempo:.4f}")
```

### Reflexión

El tiempo de ejecución no crece de forma completamente lineal al aumentar
el número de elementos. Al multiplicar por 10 el tamaño de la lista, el
tiempo aumenta más que proporcionalmente.

Esto se debe a que la función `sorted()` de Python utiliza Timsort, cuya
complejidad es O(n log n) en el peor caso. Por tanto, su crecimiento es
algo mayor que lineal.
