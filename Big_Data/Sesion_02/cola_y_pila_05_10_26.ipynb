## uso de `functools import wraps`

> functools.wraps se utiliza en Python para preservar los metadatos (como el nombre, la documentación y las anotaciones) de una función original cuando creas un decorador

### ¿Por qué se necesita?
Cuando decoras una función, la función interna (el wrapper) reemplaza a la original. Como consecuencia, la función decorada pierde su identidad: su atributo __name__ pasa a ser 'wrapper' y su documentación (__doc__) desaparece o muestra la del envoltorio. functools.wraps se encarga de copiar estos datos de la función original al wrapper.

Se aplica como un decorador dentro de tu propio decorador, pasándole como argumento la función **original**

```python
from collections import deque
import time
from functools import wraps

def medir_tiempo(func):
  @wraps(func)
  def funcion_medida(*args, **kwargs):
    ini = time.time()
    res = func(*args, **kwargs)
    fin = time.time()
    print(f"Tiempo de cómputo = {fin-ini} segundos")
    return res

  return funcion_medida


@medir_tiempo
def pushN(pila, n):
  for i in range(n):
    pila.append(i)


pila = deque()
pushN(pila, 10000)
```

## Implementación de una pila

Implementar una pila (stack) con operaciones push/pop

```python
class Pila:
  """Implementación de una pila (LIFO)"""

  def __init__(self):
    self.items = []

  def push(self, item):
    """Añadir elemento al tope - O(1)"""
    self.items.append(item)

  def pop(self):
    """Quitar y devolver el tope - O(1)"""
    if not self.is_empty():
      return self.items.pop()
    raise IndexError("Pila vacía")

  def peek(self):
    """Ver el tope sin quitarlo - O(1)"""
    if not self.is_empty():
      return self.items[-1]
    raise IndexError("Pila vacía")

  def is_empty(self):
    return len(self.items) == 0

  def size(self):
    return len(self.items)
```

## Implementación de una cola

Implementar una cola (queue) con operaciones enqueue/dequeue

```python
class cola:
  """Implementación de una cola (FIFO)"""

  def __init__(self):
    self.items = []

  def enqueue(self, item):
    """Añadir elemento al final - O(1)"""
    self.items.append(item)

  def dequeue(self):
    """Quitar y devolver el primer elemento - O(n)"""
    if not self.is_empty():
      return self.items.pop(0)
    raise IndexError("Cola vacía")

  def peek(self):
    """Ver el primer elemento sin quitarlo - O(1)"""
    if not self.is_empty():
      return self.items[0]
    raise IndexError("Cola vacía")

  def is_empty(self):
    return len(self.items) == 0

  def size(self):
    return len(self.items)
```
