# Preguntas de reflexión

## 1. ¿Qué diferencia hay entre una lista y un conjunto en términos de operaciones?

Una **lista** mantiene los elementos ordenados y permite que haya elementos repetidos. Es útil cuando importa el orden de los datos o cuando queremos acceder a ellos mediante un índice.

Un **conjunto (`set`)**, en cambio, no está pensado para mantener un orden concreto y no permite elementos duplicados. Su principal ventaja es que las operaciones de búsqueda, inserción y eliminación suelen ser más rápidas, con un coste aproximado de **O(1)**.

Por ejemplo, usaría una lista para guardar una secuencia de datos donde el orden sea importante, y un conjunto si quiero comprobar rápidamente si un elemento ya existe o eliminar duplicados.

---

## 2. ¿Por qué un diccionario (hash table) tiene búsqueda O(1)?

Un diccionario utiliza una estructura llamada **tabla hash**. Cuando guardamos una clave, se calcula mediante una función hash una posición donde almacenar ese elemento.

Cuando queremos buscar una clave, se vuelve a calcular su hash y se puede ir directamente a la posición correspondiente, en lugar de tener que recorrer todos los elementos uno por uno.

Por eso, en condiciones normales, la búsqueda, inserción y eliminación tienen un coste medio de **O(1)**. Aun así, puede haber colisiones entre claves y, en esos casos, el rendimiento puede empeorar.

---

## 3. ¿Cuándo usarías una cola vs una pila?

Usaría una **cola (queue)** cuando necesito que los elementos se procesen en el mismo orden en el que han llegado (**FIFO: First In, First Out**). Un ejemplo sería una cola de impresión o las tareas pendientes de un servidor.

Una **pila (stack)** la usaría cuando necesito que el último elemento añadido sea el primero en salir (**LIFO: Last In, First Out**). Por ejemplo, para implementar la función de "deshacer" en un programa, donde la última acción realizada es la primera que queremos deshacer.

En resumen:

* **Cola → FIFO → primero en entrar, primero en salir.**
* **Pila → LIFO → último en entrar, primero en salir.**
