# Programación orientada a objetos en Python

Python es un lenguaje de proposito general y multiparadigma. Permite, entre otros enfoques, la programación imperativa, la programación funcional y otros tipos de paradigmas, **la programación orientada a objetos (POO)**.

## Programación Orientada a Objetos (POO)

* Es un paradigma de programación que se basa en la utilización de objetos.
* Se le da más importancia a los datos que al flujo del programa, es decir, se caracteriza por dar importancia a los datos (más que al flujo del programa)
* Los datos del mundo real se modelan mediante clases. Estos datos se pasan a las clases.
* Un programa consiste en la "vida y obra" de los objetos creados a partir de esas clases: acciones de esos objetos sobre sí mismos y sobre otros objetos.


## Clase

Es un modelo que se crea a partit de una simplificación de un concepto del mundo real, a partir del cual se crean objetos. Define que características van a tener esos objetos y qué acciones tienen asociadas.

## Objeto o instancia

Deriva de la concreción de una clase. (instanciar == crear un objeto).

Una clase sería "alumno, mientras que "Ana" y "Pepito" serían objetos de esa clase.

Los datos asociados a cada objeto se almacenan en los **atributos de instancia** (diferente a atributos de clase). Por ejemplo, cada coche puede tener su propia matrícula, marca, modelo o color.

"Así funcionan los ORM"

Las acciones asociadas a los objetos se definen mediante **métodos de instancia** (no de clase). Por ejemplo, un coche podría arrancar, detenerse o recorrer una distancia.


## `self` y `__init__()`

En los métoddos de instancia, `self` representa el objeto sobre el que se está trabajando.

El método especial `__init__()` sería un inicializador y se ejecuta inmediatamente despúes de crear un objeto y se utiliza habitualmente para inicializar sus atributos. (es como el constructor de Java, pero en Python no es exactamente igual al constructor).

El constructor real es `__new_()` pero en Python se usa en casos muy puntuales.


```python
class GatoBasico:
  """(docString)Representa un gato de forma muy simplificada"""

  def __init__(self, n, c="marrón", e=1):  # parámetros (normalmente se llaman igual a los atributos), con el igual se le dan valores por defecto a los parámatros si no ponemos niguno al instanciar
    # atributos
    self.nombre = n
    self.color = c
    self.edad = e

  # métodos
  def maulla(self):
    print("Miauu")

  def ronronea(self):
    print("mrrrrrr")

  # para hacer métodos que devulven true o false
  #def esta_casado():
```

```python
# crear un objeto (instanciar)
garfield = GatoBasico("Garfield", "naranja", 10)
lisa = GatoBasico("Lisa")

print(garfield.nombre, garfield.color, garfield.edad)
print(lisa.nombre, lisa.color, lisa.edad)
```

## El método `__str__()`

Este método especial permite definir una representación textual legible de un objeto. Es el equivalente al método `toString()` de Java

```python
from enum import Enum
import random

# definir una clase como subclase de Enum
class Sexo(Enum):
  MACHO = "macho"
  HEMBRA = "hembra"

# clase Gato más completa
class Gato:
  """(docString)Representa un gato de forma más completa"""

  def __init__(
        self,
        nombre: str,
        color: str = "marrón",
        sexo: Sexo = Sexo.HEMBRA,
        raza: str = "mestizo",
        edad: int = 1,
        peso: float = 1.5,
        energia: int = 50,
        fuerza: int = 100
      ) -> None: # esto significa que la clase no devuelve nada, esto especifica el tipo de dato que devuelve la clase
    self.nombre = nombre
    self.color = color
    self.sexo = sexo
    self.raza = raza
    self.edad = edad
    self.peso = peso
    self.energia = energia
    self.fuerza = fuerza

  # métodos especiales se escriben juntos
  # el método str da información del objeto
  def __str__(self) -> str:
    gatx = "una gata" if self.sexo == Sexo.HEMBRA else "un gato"
    return (
      f"Hola, me llamo {self.nombre}, soy {gatx}, "
      f"de raza {self.raza} y de color {self.color}, "
      f"mi edad es {self.edad} años y peso {self.peso}kg."
    )

  # métodos
  def maulla(self) -> None:
    print("Miauu")

  def ronronea(self) -> None:
    print("mrrrrrr")

  def come(self, comida: str) -> None:
    """Acepta pescado y rechaza cualquier otra comida"""
    if comida.lower() == "pescado":
      print(f"{self.nombre} dice: El pescado está muy bueno")
    else:
      print(f"{self.nombre} dice: No me gusta comer {comida}, yo solo como pescado")

  # método que reciba como parámetro otro objeto de la misma clase
  def saluda_a(self, otro: "Gato") -> str:
    """Hace que un gato salude a otro"""
    saludo: str
    saludo = f"{self.nombre} frota la cabeza contra {otro.nombre}"
    return saludo

  def pelea_con(self, rival: "Gato") -> None:
    """Simula una pelea entre dos gatos.
    Solo hay pelea si ambos tienen al menos 10 puntos de energía.
    El gato con más fuerza gana y tras la pelea ambos pierden energía
    y ganan algo de fuerza.
    Por defecto fuerza 100 y energia 50
    Pérdida de energia: entre 1 y 5
    Ganancia de fuerza: entre 1 y 10
    """

    if self.energia > 10 and rival.energia > 10:
      print(f"{self.nombre} pelea contra {rival.nombre}")

      if self.fuerza > rival.fuerza:
        print(f"{self.nombre} gana la pelea contra {rival.nombre}")
      elif self.fuerza < rival.fuerza:
        print(f"{rival.nombre} gana la pelea contra {self.nombre}")
      else:
        print("Empate")

      self.energia = self.energia - random.randint(1, 5)
      self.fuerza = self.fuerza + random.randint(1, 10)
      rival.energia = rival.energia - random.randint(1, 5)
      rival.fuerza = rival.fuerza + random.randint(1, 10)

    elif self.energia == rival.energia:
      print("Empate, ambos tienen la misma energía")

    else:
      if self.energia < 10:
        print(f"{self.nombre} tiene menos de 10 puntos de energía. No hay pelea.")
      elif rival.energia < 10:
        print(f"{rival.nombre} tiene menos de 10 puntos de energía. No hay pelea.")
      else:
        print("Los dos tienen menos de 10 puntos de energía. No hay pelea")
```

```python
garfield = Gato("Garfield", color="naranja", sexo=Sexo.MACHO, raza="naranja", edad=6, peso=10)
lisa = Gato("Lisa")

print(garfield)
print(lisa)

print("\nSe le da de comer")
garfield.come("pescado")
lisa.come("carne")

print("\n")
garfield.saluda_a(lisa)
```

```python
print("Primera pelea:\n----------")
garfield.pelea_con(lisa)
print("Segunda pelea\n----------")
garfield.pelea_con(lisa)
```

## Atributos y métodos de clase

Los atributos de instancia pertencen a cada objeto.
Por ejemplo, cada vehículo tendrá su propio kilometraje.

Un **atributo de clase**, sin embargo, pertenece a la clase y es compartido por sus instancias. Es útil para almacenar información común, como el kilometraje acumulado de todos los vehículos. Por ejemplo una aplicación de compraventa de coches donde se quiere saber el kilometraje de cada vehículo individual, pero también se quiere saber el kilometraje total de todos los vehículos.

Los **Métoodo de clase** en vez de aplicarse a un objeto, se le aplica a la clase. Por ejemplo, mostrar info del concesionario. Suele hacer uso del atributo de clase.

Un método decorado con `@classmethod` recibe la clase como primer argumento, que por convenio es `cls`

```python
class Vehiculo:
  """Representa un vehículo y su kilometraje"""

  kilometraje_total: int = 0

  def __init__(self, marca: str, modelo: str) -> None:
    self.marca: str = marca
    self.modelo: str = modelo
    self.kilometraje: int = 0

  def __str__(self) -> str:
    return (
        f"{self.marca}, {self.modelo} "
        f"con kilometraje {self.kilometraje} km"
    )

  # métodos
  def recorre(self, km: int) -> None:
    """Incrementa el kilometraje del vehículo individual y toal de todos (km acumulado)"""
    if km < 0:
      raise ValueError("La distancia no puede ser negativa.")

    self.kilometraje += km
    Vehiculo.kilometraje_total += km

  # método de clase, para no tener que llamar a un objeto creado con este método que es global,
  # ejemplo: vehiculo3.muestra_kilometraje_total()
  # cls es porque se usa para la clase, por eso no se usa self que sería para un objeto
  @classmethod
  def muestra_kilometraje_total(cls):
    print(f"Entre todos los vehículos han recorrido {cls.kilometraje_total} km.")
```

```python
vehiculo1 = Vehiculo("Cupra", "Raval")
print(vehiculo1)
vehiculo2 = Vehiculo("Seat", "León")
vehiculo3 = Vehiculo("Honda", "PCX")

vehiculo1.recorre(500)
vehiculo2.recorre(200)
vehiculo3.recorre(75)

print(vehiculo1)
print(vehiculo2)
print(vehiculo3)
print(f"Kilometraje total: {Vehiculo.kilometraje_total} km")

vehiculo2.recorre(100)
print(Vehiculo.muestra_kilometraje_total())
```
