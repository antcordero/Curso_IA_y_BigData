# Map, Filter y Reduce

Las funciones `map()`, `filter()` y `reduce()` son extremadamente potentes y, bien utilizadas, ayuda a reducir las líneas de un programa y a dejar el código más limpio.

Están basadas en el paradigma de la programación funcional.

Las tres se aplican recorriendo estructuras iterables, por ejemplo listas, y realizando determinadas acciones sobre esos iterables.

## Funciones de orden superior (higher-order functions)

Una función de orden superior es aquella que recibe otra función como parámetro o devuelve una función como resultado.

````python
def decora_con_estrellas(texto):
  return f"⭐⭐⭐ {texto} ⭐⭐⭐"

def triplica_texto(texto):
  return (texto + " ") * 3

def no_hace_nada(texto):
  return texto

# Función de orden superior
def saluda(funcion, nombre):
  print(funcion(f"Hola {nombre}!"))
````

````python
saluda(decora_con_estrellas, "Toni")
````

````python
saluda(triplica_texto, "Toni")
````

````python
saluda(no_hace_nada, "Toni")
````

````python
def crear_multiplicador(n):
  """Si se le pasa una función que sea un multiplicador de un número"""

  def multiplicar(x):
    return x * n

  return multiplicar  # Devuelve una función
````

````python
duplica = crear_multiplicador(2)
triplica = crear_multiplicador(3)
````

````python
print(duplica(100))
print(triplica(500))
````

## Funciones lambda

Son expresiones que permiten definir funciones anónimas con un formato muy conciso.

Una función lambda solo puede contener una expresión. Si necesitamos varias sentenias o una lógica más completa, es mejor recurrir a las funciones tradicionales.

````python
# una función que diga si un número es par (tradicional)
def es_par(n):
  return (n % 2 == 0)

print(f"Es par? (forma tradicional): {es_par(7)}")
print(f"Es par? (forma tradicional): {es_par(22)}")
````

````python
# forma lambda
par = lambda n: (n % 2) == 0

print(par(8))
print(par(9))
````

## Map

Transforma cada uno de los elementos del iterable.

````python
def cubo(n):
  return n ** 3

numeros = [5, 7, 22, 97, 54, 62, 77, 23, 73, 61]

# primero va la funcion y luego el iterable
numeros_al_cubo = list(map(cubo, numeros)) # map solamente no da una lista, da una estructura de donde se puede ir sacando de uno en uno

print(numeros)
print(numeros_al_cubo)
````

````python
# ahora ejemplo con una lambda

numeros_al_cubo = list(map(lambda n: n ** 3, numeros))

print(numeros)
print(numeros_al_cubo)
````

En Python, `map()` devuelve un iterador, no una lista.

````python
numeros_al_cubo = map(lambda n: n ** 3, numeros)
print(numeros_al_cubo)
````

## Ejercicio:

Utiliza un `map` para pasar todos los elementos de una lista a mayúsculas

````python
frutas = ["manzana", "pera", "melocotón", "chirimoya", "caqui", "ciruela", "naranja", "mango"]

def poner_mayuscula(palabra):
  # existe esto en Python: palabra.capitalize()
  return (palabra[0].upper() + palabra[1:])

frutas_mayusculas = list(map(poner_mayuscula, frutas))

print(frutas)
print(frutas_mayusculas)
````

## Filter

````python
# Filtrar los números positivos (mayores que cero)
numeros = [-1, 5, -7, 22, 97, 0, 54, 62, -77, 23, 73, -61]

print(numeros)
positivos = list(filter(lambda x: x > 0, numeros))
````

## Reduce

La función `reduce` sirve para consumir los elementos de un iterable mediante una función y obtener un valor.

Se obtiene como resultado un único elemento que puede ser de cualquier tipo.

La función que se utiliza para consumir elementos tiene dos parámetros, el primero es el acumulado de lo que se lleva consumiendo y el segundo es el elemento que se va sacando.

El formato es el siguiente:

```
reduce(funcion_de_consumo, iterable, valor_inicial)
```

````python
from functools import reduce

numeros = [1, 5, 7, 22, 97, 0, 54, 62, 77, 23, 73, 61]

suma = reduce(lambda acumulado, elemento: acumulado + elemento, numeros, 0)
# función de consumo: toda la lambda
# iterable: numeros
# valor inicial: 0

print(suma)
````

## Combinando `map`, `filter` y `reduce`

````python
# Suma de los cuadrados de los números pares

numeros = [1, 2, 3, 4, 5, 6]

pares = filter(lambda x: x % 2 == 0, numeros)
cuadrados = map(lambda x: x ** 2, pares)
suma = reduce(lambda acumulado, elemento: acumulado + elemento, cuadrados, 0)

print(suma)
````

## Ejercicio

Dada la siguiente lista de productos, utiliza `filter`, `map` y `reduce` para calcular el valor total de los productos que quedan unidades.

```
productos = [
    {"nombre": "Ratón", "precio": 15, "stock": 10},
    {"nombre": "Teclado", "precio": 25, "stock": 0},
    {"nombre": "Monitor", "precio": 180, "stock": 4},
    {"nombre": "USB", "precio": 8, "stock": 20}
]
```

````python
from functools import reduce

productos = [
    {"nombre": "Ratón", "precio": 15, "stock": 10},
    {"nombre": "Teclado", "precio": 25, "stock": 0},
    {"nombre": "Monitor", "precio": 180, "stock": 4},
    {"nombre": "USB", "precio": 8, "stock": 20}
]

productos_con_unidades = filter(lambda p: p["stock"] > 0, productos)
valor_stock = map(lambda p: p["precio"] * p["stock"], productos_con_unidades)
valor_total_stock = reduce(lambda acumulado, valor: acumulado + valor, valor_stock, 0)

print(valor_total_stock)
````
