## Variables

Por ejemplo sería correcto escribir `precio_sin_iva`.
Y no sería correcto `precioSinIva`.


```python
x = 88
print(x)
print(type(x))
```

```python
pi = 3.141592654
print(pi)
print(type(pi))
```

## Operadores

```python
print(x + pi)
print(x * pi)
print(x ** 2)
dos_frutas = fruta + fruta
frutero = fruta * 5
print(dos_frutas)
print(frutero)
```

```python
print(7 / 2)  # división con decimales
print(7 // 2) # división entera
print(7 % 2)  # resto de la división (módulo)
```

```python
print(f"La variable x multiplicada por pi vale {x * pi:.2f}.") #cadena formateada
```

## Listas

El índice, como en la mayoría de los lenguajes de programación, comienza en 0.

```python
frutas = ["manzana", "pera", "melocotón", "chirimoya", "caqui", "ciruela", "naranja", "mango"]
```

```python
print(frutas)
```

```python
len(frutas)
```

```python
frutas[5]
```

```python
#sacar un trozo de un array dependiendo de la posición que queramos, por ejemplo de la 2 a la 7
frutas[1:3] # coge desde el primer índice hasta uno antes del segundo índice que se indica
```

```python
# para coger desde melocotón a naranja
frutas[2:7]
```

```python
# coger mango
frutas[7] #opción sabiendo la posición
```

```python
# coger el último elemento, cuando no sabemos la longitud de la lista
frutas [-1]
```

```python
frutas[-3]
```

```python
frutas[-1:-4] # no imprime porque primero busca desde la posición de la izquierda y sigue hacia derecha
```

```python
frutas[6:2]
```

```python
frutas [-4:-1] # posición de la izq hasta la posición de la derecha pero la anterior
```

```python
frutas[1:-2]
```

```python
frutas[3:]  # sin índice final va hasta el final de la lista
```

```python
frutas[:4]  # desde el principio hasta la posición
```

```python
# estas reglas también sirven para strings
fruta[:4]
```

```python
# saber si un string está dentro de la lista
"caqui" in frutas
```

```python
"melón" in frutas
```

```python
# ver el índice de un elemento
frutas.index("caqui")
```

```python
# en las listas puede haber elementos repetidos y de distinto tipo
cajon_desastre = [22, True, 7.8, "hola"]
cajon_desastre
```
