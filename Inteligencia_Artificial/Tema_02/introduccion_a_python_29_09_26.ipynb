```python
# estructura parecida al switch - case
dia = input("Dime el día de la semana (por ej. lunes):").lower()

cadena = "Ese día, a primera hora toca "

match dia:
  case "lunes":
    print(cadena + "Inteligencia Artificial.")
  case "martes":
    print(cadena + "Inteligencia Artificial.")
  case "miércoles":
    print("¡Día libre!")
  case "jueves":
    print(cadena + "Big Data.")
  case "viernes":
    print(cadena + "Inteligencia Artificial.")
  case "sábado":
    print("¡Día libre!")
  case "domingo":
    print("¡Día libre!")
  case _:
    print("Ese día no existe.")
```

```python
# se puede simplificar
# estructura parecida al switch - case
dia = input("Dime el día de la semana (por ej. lunes):").lower()

cadena = "Ese día, a cuarta hora toca "

match dia:
  case "lunes" | "martes":
    print(cadena + "Big Data.")
  case "martes":
    print(cadena + "Big Data.")
  case "miércoles" | "sábado" | "domingo":
    print("¡Día libre!")
  case "jueves":
    print(cadena + "Inteligencia Artificial.")
  case "viernes":
    print(cadena + "Ya terminaste.")
  case _:
    print("Ese día no existe.")
```

```python
# El range
# para Java es el for y la condición es: for (INICIO; CONDICIÓN; CAMBIO)
# para Python se hace con el range y es: range(INICIO, FINAL, PASO)
```

```python
def piramide(altura):
  fila = 1
  for espacios in range(altura -1, 0, -1):
    if fila == 1:

      for i in range(espacios):
        print(" ", end="")
      print("*")

    else:

      for j in range(espacios):
        print(" ", end="")
      print("*", end="")

      for a in range(fila*2-3):
        print(" ", end="")
      print("*")

    fila += 1
```

```python
piramide(6)
```

```python
#piramide llena (solución de clase)
def piramide_llena(altura):
  for i in range(altura,0,-1):
    print(" " * (i-1) + "*" * (1+(altura-i) * 2))

print(piramide_llena(5))
```

```python
#Otra solución:
def piramide_llena2(altura):
  for i in range(altura):
      espacios = " " * (altura - i)
      asteriscos = "*" * (2 * i + 1)
      print(espacios + asteriscos)

print(piramide_llena2(5))
```
