# List Comprehension (comprensión de listas)

La comprensión de listas es una forma muy concisa de crear listas.


Se podría hacer con bucles, pero de esta forma queda más conciso.

La sintáxis básica es la siguiente:

```python
lista_nueva = [expresion for elemento in iterable if condicion]
```

Para la creación de la lista se ejecutan los siguientes pasos:

1. `for elemento in iterable`: Itera sobre cada elemento del iterable que (casi siempre va a ser una lista)
2. `if condicion`: Filtra los elementos que cumplen una condición.
3. `expresion`: Aplica esta expresión a cada elemento filtrado.


````python
# Crea una lista de los cuadrados del 0 al 9

cuadrados = [n ** 2 for n in range(10)]
print(cuadrados)
````

````python
# crea una lista de los cuadrados de los números pares del 0 al 9
cuadrados_de_pares = [n ** 2 for n in range(10) if n%2==0]
print(cuadrados_de_pares)
````

````python
# Crea una lista con las palabras en mayúsculas que tienen una longitud mayor que 5
frutas = ["manzana", "pera", "melocotón", "chirimoya", "caqui", "ciruela", "naranja", "mango"]

mayusculas = [fruta.upper() for fruta in frutas if len(fruta) > 5]

print(mayusculas)
````

````python
# A partir del diccionario de la brisca, crea una lista con los nombres de las figuras que tienen algún valor
puntos_brisca = {
  "as": 11,
  "dos": 0,
  "tres": 10,
  "cuatro": 0,
  "cinco": 0,
  "seis": 0,
  "siete": 0,
  "sota": 2,
  "caballo": 3,
  "rey": 4,
}

figuras_con_valor = [figura for figura in puntos_brisca if puntos_brisca[figura] > 0]

print(figuras_con_valor)
````

````python
# otra forma de hacerlo, esta solución es más pythonic (más acorde al estilo de python)
figuras_con_valor2 = [figura for figura, valor in puntos_brisca.items() if valor > 0]
print(figuras_con_valor2)
````

## Condicionales con `if-else` en la expresión

Cuando queremos transformar un elemento en función de una condición (y no solo filtrarlo), la estructura `if-else` se coloca antes del `for`:

```python
lista_nueva = [expresion_si_true if condicion2 else expresion_si_false for elemento in iterable if condicion1]
```

Ejemplo:

````python
# Crea una lista a partir de otra reemplazando los números impares por 0 y manteniendo los pares

numeros = list(range(10))
pares_o_cero = [n if n%2==0 else 0 for n in numeros]
print(pares_o_cero)
````

## Comprensiones anidadas
Comprensión de listas anidadas
Podemos utilizar más de un bucle `for` dentro de la comprensión.

````python
# Genera una baraja (combinaciones de palo con figura)

palos = ["oros", "copas"]
figuras = ["as", "tres", "sota", "caballo", "rey"]

baraja = [f"{figura} de {palo}" for palo in palos for figura in figuras]
print(baraja)
````

## Compresiones de diccionarios y conjuntos

La misma sintaxis concisa se puede aplicar a otro tipo de iterables

````python
# Comprensión de diccionarios (dictionary comprehension)
cuadrados_dict = {x: x ** 2 for x in range(10)}
print(cuadrados_dict)
````

````python
# Comprensión de conjuntos (set comprehension)
#set(list("abracadabra"))
letras = {letra.upper() for letra in "abracadabra"}
letras
````

## Buenas prácticas con `list comprehension`

1. **Legibilidad primero**: Si la lista de comprensión sale demasioado larga o compleja, es preferible usar un bucle `for` tradicional.
2. **Uso de la memoria**: Para conjunto de datos extremadamente grandes, es más eficiente usar **expresiones generadoras** sustituyendo corchetes por paréntesis.

````python
import sys

# Lista en memoria (1 millón de elementos)
lista = [x for x in range(1_000_000)]

# Generador (mismo número de elementos, pero bajo demanda)
generador = (x for x in range(1_000_000))

print(f"Tamaño de la lista en RAM: {sys.getsizeof(lista)} bytes")
# Muestra aprox: ~8.000.000 bytes (~8 MB)

print(f"Tamaño del generador en RAM: {sys.getsizeof(generador)} bytes")
# Muestra aprox: ~200 bytes
````
